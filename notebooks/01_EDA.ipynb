# 01 - Exploratory Data Analysis (EDA)
**AI-Based Fake News & Misinformation Detection**

This notebook explores the dataset used for training the detection pipeline:
- Class balance (Real vs Fake)
- Article text length distributions
- Distinctive linguistic characteristics


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# Load processed datasets
train_df = pd.read_csv('../data/processed/train.csv')
test_df = pd.read_csv('../data/processed/test.csv')
print(f'Training samples: {len(train_df)}, Testing samples: {len(test_df)}')
train_df.head(3)
```

```python
# Class distribution
print('Class distribution in training set:')
print(train_df['label'].value_counts())

train_df['label'].value_counts().plot(kind='bar', color=['#3b82f6', '#ef4444'], title='Class Distribution')
plt.ylabel('Count')
plt.show()
```

```python
# Length statistics
print('Summary of word counts:')
print(train_df.groupby('label')['word_count'].describe())
```

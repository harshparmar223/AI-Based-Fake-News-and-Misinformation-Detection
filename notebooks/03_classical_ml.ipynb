# 03 - Machine Learning Model Baselines & Evaluation
Comparative evaluation of SVM, Random Forest, and Logistic Regression models.


```python
import json
import pandas as pd

with open('../models/evaluation_results.json') as f:
    metrics = json.load(f)

df = pd.DataFrame(metrics).T
print('Model Performance Comparison:')
print(df[['accuracy', 'precision', 'recall', 'f1', 'roc_auc']])

df[['accuracy', 'f1', 'roc_auc']].plot(kind='bar', figsize=(8, 4), title='Model Performance Comparison')
```

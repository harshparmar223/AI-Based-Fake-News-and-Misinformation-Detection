# 04 - Explainable AI (LIME) and Uncertainty
Interpreting predictions at token level and quantifying uncertainty.


```python
import sys
sys.path.append('..')
from src.ml.predictor import EnsemblePredictor
from src.explainability.lime_explainer import LIMEExplainer
from src.explainability.uncertainty import compute_uncertainty

predictor = EnsemblePredictor.load_default()
explainer = LIMEExplainer(num_features=6, num_samples=150)

sample = 'Government secret project involving artificial weather control revealed by whistleblowers.'
pred = predictor.predict(sample)
print(f'Prediction: {pred.label} (Confidence: {pred.confidence:.1%})')

unc = compute_uncertainty(pred.confidence, pred.fake_probability, pred.model_scores, sample)
print(f'Uncertainty: {unc.level} (Score: {unc.uncertainty_score})')

words = explainer.explain_structured(sample, predictor.predict_proba)
print('LIME Word Attributions:')
for w in words:
    print(f"  {w['word']:<15} {w['weight']:>+7.4f} ({w['influence']})")
```
